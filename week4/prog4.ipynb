```python
#list operations
#concatenation
l1=[1,2,3,4,5]
l2=[7,8,9,7,10]
print(l1+l2)
```

```python
#repetition
print(l1*4)
```

```python
#indexing
print(l1[2])
```

```python
#slicing
print(l1[1:5])
print(l1[2:])
print(l1[:3])
```

```python
#membership operators
print(6 in l1)
```

```python
#iteration
for i in range(len(l1)):
  print(l1[i])
```

```python
#built in functions
#len
print(len(l1))
```

```python
#sum
print(sum(l2))
```

```python
#min
print(min(l2))
```

```python
#max
print(max(l1))
```

```python
#reversed
print(list(reversed(l2)))
```

```python
#sorted
print(sorted(l2))
```

```python
#list methods
#append
l1.append(29)
print(l1)
```

```python
#insert
l1.insert(5,60)
print(l1)
```

```python
#pop
l1.pop()
print(l1)
```

```python
#remove
l1.remove(60)
print(l1)
```

```python
#sort
l1.sort
print(l1)
```

```python
#reverse
l1.reverse()
print(l1)
```

```python
#count
print(l1.count(4))
```

```python
#index
print(l1.index(1))
```

```python
#extend
l1.extend(l2)
print(l1)
```

```python
#list comprehension
l=list(map(int,input().split()))
print(l)
```

```python
l=[i**3 for i in range(7)]
print(l)
```

```python
l=[int(i) for i in input().split()]
print(l)
```
