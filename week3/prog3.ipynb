```python
#prints reverse of numbers
def reverse_numbers(n):
  reverse=0

  while n>0:
    digit=n%10
    reverse=reverse*10+digit
    n=n//10
  return reverse

n=int(input())
reverse=reverse_numbers(n)
print(reverse)
```
