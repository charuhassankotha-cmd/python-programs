```python
#arithmetic operators
x=int(input("enter value of x:"))
y=int(input("enter value of y:"))
print("sum of x,y is",x+y)
print("sub of x,y is",x-y)
print("mul of x,y is",x*y)
print("div of x,y is",x/y)
print("mod of x,y is",x%y)
print("floor div of x,y is",x//y)
print("exp of x,y is",x**y)
```

```python
#comparisson operators
x=int(input("enter value of x:"))
y=int(input("enter value of y:"))
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x>=y)
print(x<=y)
```

```python
#logical operators
x=int(input("enter value of x:"))
y=int(input("enter value of y:"))
print(x>y and x<y)
print(x>y or x<y)
print(not(x>y and x<y ))
```

```python
#membership operators
x=str(input("enter the value of x: "))
y=str(input("enter the value of y: "))
print(x is y)
print(x is not y )
```

```python
#identity operators
x=int(input("enter value of x:"))
y=int(input("enter value of y:"))
print(x is y)
print(x is not y)
```

```python
#bitwise operators
x=int(input("enter value of x:"))
y=int(input("enter value of y:"))
print(x>y & x<y)
print(x>y | x<y)
print(x>y ^ x<y)

```
