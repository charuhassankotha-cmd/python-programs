```python
# to find duplication
l=[int (i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
```

```python
# program to find non repeated elements
l=[int (i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)==1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
```

```python
# program that prints missing numbers  between 1 to n
l=[int (i) for i in input().split()]
l2=[]
n=len(l)
for i in range(l[0],l[n-1]+1):
  if(i not in l):
    l2.append(i)
print(l2)
s=set(l2)
print(s)
```
